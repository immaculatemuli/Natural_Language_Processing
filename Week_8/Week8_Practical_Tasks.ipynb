This notebook covers Practical Task 1 (Sentiment Analysis using Hugging Face Transformers) and Practical Task 2 (Text Generation using GPT-2).

```python
classifier = pipeline('sentiment-analysis')

print('Sentiment analysis pipeline loaded (DistilBERT fine-tuned on SST-2).')
```

```python
test_statements = {
    'Positive Review': 'The lecturer explained the topic so clearly, I finally understand Transformers.',
    'Negative Review': 'The lecture hall was too crowded and the projector kept failing.',
    'Neutral Statement': 'The class started at nine in the morning and ended at eleven.'
}

print('SENTIMENT ANALYSIS RESULTS')
print('=' * 65)

for label, text in test_statements.items():
    result = classifier(text)[0]
    print()
    print(f'{label}:')
    print(f'  Text:       "{text}"')
    print(f'  Prediction: {result["label"]}')
    print(f'  Confidence: {result["score"]:.2%}')
```

### Interpretation of Results

The positive review was correctly classified as POSITIVE with high confidence because words like "clearly" and "understand" carry positive meaning. The negative review was correctly classified as NEGATIVE because phrases like "too crowded" and "kept failing" signal dissatisfaction. The neutral statement is interesting because this particular model only outputs POSITIVE or NEGATIVE — it has no NEUTRAL category, so a purely factual statement still gets pushed toward one label, usually with a lower confidence score than the clearly positive or negative examples. This shows a limitation of pretrained sentiment models: they are only as good as the labels they were trained on.

## Practical Task 2 Text Generation using GPT-2

**Requirements:**
- Accept user input
- Generate text continuations
- Display generated text
- Experiment with different prompt sentences and compare outputs

```python


# STEP 1: Install and import the required library
# transformers = Hugging Face's library that gives us easy access
# to pretrained AI models like GPT-2 without building one ourselves.
!pip install transformers torch --quiet

from transformers import pipeline

print('Loading GPT-2... this may take a minute the first time.')

# "pipeline" is a shortcut function — it loads the model, the
# tokenizer (which turns words into numbers GPT-2 understands),
# and wraps everything into one easy-to-use object.
generator = pipeline('text-generation', model='gpt2')

print('GPT-2 is ready!')
print()


# STEP 2: A function that takes your prompt and returns GPT-2's
#            continuation. Keeping this as a function means we can
#            reuse it for both the interactive part AND the
#            comparison part further down.
def generate_text(prompt, max_length=50):
    """
    Takes a sentence (prompt) and returns GPT-2's continuation.

    max_length = the total number of words/tokens GPT-2 is allowed
    to produce, INCLUDING your original prompt.
    """
    result = generator(
        prompt,
        max_length=max_length,
        num_return_sequences=1,   # only generate ONE version of the continuation
        truncation=True
    )
    # result is a list containing a dictionary; we want the text inside it
    return result[0]['generated_text']


# STEP 3: ACCEPT USER INPUT and display generated text
# This loop keeps asking you for prompts until you type "quit".
# Requirement 1 (accept user input) + Requirement 3 (display output)
# both happen right here.

print('=' * 65)
print('  GPT-2 TEXT GENERATOR')
print('  Type any sentence and GPT-2 will continue it for you.')
print('  Type "quit" when you want to stop.')
print('=' * 65)

while True:
    print()
    user_prompt = input('YOUR PROMPT: ').strip()

    # Exit condition — stops the loop if you type quit/exit/nothing
    if user_prompt.lower() in ['quit', 'exit', '']:
        print('\nGoodbye! Text generation stopped.')
        break

    # Requirement 2: generate the text continuation
    generated = generate_text(user_prompt)

    # Requirement 3: display the generated text clearly
    print()
    print(f'  YOU TYPED:      "{user_prompt}"')
    print(f'  GPT-2 GENERATED: "{generated}"')
    print('-' * 65)


# STEP 4: EXPERIMENT WITH DIFFERENT PROMPTS AND COMPARE

print()
print('=' * 65)
print('  COMPARING DIFFERENT TYPES OF PROMPTS')
print('=' * 65)

comparison_prompts = [
    'The weather today is',                      # simple, everyday sentence
    'Artificial Intelligence will',               # open-ended, futuristic topic
    'My favorite subject in school is',           # personal/opinion-style prompt
    'Once upon a time, there was a',              # storytelling style
]

for prompt in comparison_prompts:
    output = generate_text(prompt, max_length=40)
    print()
    print(f'PROMPT:    "{prompt}"')
    print(f'GENERATED: "{output}"')

print()
print('=' * 65)
print('  WHAT TO NOTICE WHEN COMPARING THE OUTPUTS ABOVE')
print('=' * 65)
print('''
  - Everyday prompts ("The weather today is") tend to get fairly
    sensible, grounded continuations because GPT-2 has seen this
    exact pattern thousands of times in training data.

  - Open-ended prompts ("Artificial Intelligence will") can produce
    more unpredictable continuations, since there are many possible
    directions GPT-2 could take the sentence.

  - Storytelling prompts ("Once upon a time...") almost always shift
    GPT-2 into a narrative style, because that phrase is strongly
    associated with story text in its training data.

  - GPT-2 generates ONE WORD AT A TIME based on probability. It has
    no real "plan" for the whole sentence, which is why longer
    outputs can sometimes drift off-topic or repeat themselves.
''')
```
